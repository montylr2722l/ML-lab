```python
# Import Libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


# Input dataset

X = np.array(([2, 9], [1, 5], [3, 6]), dtype=float)

y = np.array([1, 2, 3])
y = y.T

# Normalize input
X = X / np.amax(X, axis=0)
X = np.round(X, 2)

X

```

```python
# Sigmoid activation function

def sigmoid(x):
    return 1 / (1 + np.exp(-x))



# Network structure

input_neurons = 2
hidden_neurons = 3
output_neurons = 1

```

```python
# Initialize weights & biases

wh = np.random.uniform(
    size=(input_neurons, hidden_neurons)
)

wh = np.round(wh, 2)

bh = np.random.uniform(
    size=(1, hidden_neurons)
)

bh = np.round(bh, 2)

wout = np.random.uniform(
    size=(hidden_neurons, output_neurons)
)

wout = np.round(wout, 2)

bout = np.random.uniform(
    size=(1, output_neurons)
)

bout = np.round(bout, 2)

print("Input Weight:\n", wh)
print("Input bias:\n", bh)

print("Output Weight:\n", wout)
print("Output bias:\n", bout)
```

```python
# ----- FORWARD PROPAGATION -----

# Hidden Layer Calculation

hidden_input = np.dot(X, wh) + bh

hidden_output = sigmoid(hidden_input)


# Output Layer Calculation

final_input = np.dot(hidden_output, wout) + bout

predicted_output = sigmoid(final_input)
```

```python
# Display results

print("Input:\n", X)

print("\nHidden Layer Output:\n", hidden_output)

print("\nPredicted Output:\n", predicted_output)
```

```python
# ----- ERROR CALCULATION -----

# Error
error = y - predicted_output

# Mean Squared Error
mse = np.mean(np.square(error))

mse
```

# Backpropagation

```python
# Input dataset

X = np.array(([2, 9], [1, 5], [3, 6]), dtype=float)

y = np.array(([92], [86], [89]), dtype=float)

# Normalize data

X = X / np.amax(X, axis=0)

y = y / 100
```

```python
# Sigmoid activation function

def sigmoid(x):
    return 1 / (1 + np.exp(-x))


# Derivative of sigmoid

def derivatives_sigmoid(x):
    return x * (1 - x)
```

```python
# Hyperparameters

epoch = 7000

lr = 0.1
```

```python
# Network architecture

input_neurons = 2
hidden_neurons = 3
output_neurons = 1
```

```python
# Weight & bias initialization

wh = np.random.uniform(
    size=(input_neurons, hidden_neurons)
)

bh = np.random.uniform(
    size=(1, hidden_neurons)
)

wout = np.random.uniform(
    size=(hidden_neurons, output_neurons)
)

bout = np.random.uniform(
    size=(1, output_neurons)
)
```

```python
# Training

for i in range(epoch):

    # Forward propagation

    hidden_input = np.dot(X, wh) + bh

    hidden_output = sigmoid(hidden_input)

    final_input = np.dot(hidden_output, wout) + bout

    output = sigmoid(final_input)


    # Error calculation

    EO = y - output


    # Backpropagation

    d_output = EO * derivatives_sigmoid(output)

    EH = d_output.dot(wout.T)

    d_hidden = EH * derivatives_sigmoid(hidden_output)


    # Weight updates

    wout += hidden_output.T.dot(d_output) * lr

    wh += X.T.dot(d_hidden) * lr

    bout += np.sum(
        d_output,
        axis=0,
        keepdims=True
    ) * lr

    bh += np.sum(
        d_hidden,
        axis=0,
        keepdims=True
    ) * lr
```

```python
# Final Output

print("Input:\n", X)

print("Actual Output:\n", y)

print("Predicted Output:\n", output)

# Error

error = y - output

# Mean Squared Error

mse = np.mean(np.square(error))

mse
```
