```python
#import required libraries

import numpy as np 
import pandas as pd
from scipy import stats
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
```

```python

# -------------------------------------------------
# 1. DATA CREATION USING NUMPY
# -------------------------------------------------

data = np.array([
    [20, 2, 55],
    [21, 4, 70],
    [20, 6, 82],
    [22, 8, 95],
    [21, 5, 76]
])

print("Data created using NumPy:")
print(data)

```

```python
# convert data into pandas dataframe


df = pd.DataFrame(
    data, columns = ["age","study-hours","marks"]
)
print("\nDataFrame:")
print(df)
```

```python
# data inspection 

print("\n first 5 rows:")
print(df.head())

print("\n dataset information")
print(df.info())

print("\nStatistical summary:")
print(df.describe())
```

```python
#checking missing values

print("\n missing values")
print(df.isnull().sum())
```

```python
#scaling using scikit learn

scaler = StandardScaler()

scaled_data = scaler.fit_transform(df)
scaled_df = pd.DataFrame(
    scaled_data, columns = df.columns 
)
print("\n scaled data:")
print(scaled_df)
```

```python
#sci-py z-score

z_scores = stats.zscore(df["marks"])
print("\nZ-sclres of marks:")
print(z_scores)
```

```python
#visualization 

plt.scatter(df["study-hours"], df["marks"])
plt.xlabel("study-hours")
plt.ylabel("marks")
plt.title("study hours vs marks")
plt.show()
```

```python
#export data to csv

df.to_csv("student_data.csv", index = False)
print("\n data exported successfully to student_data.csv")
```
